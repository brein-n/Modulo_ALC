Implementar la función obtenerQ2d(imgPts, objPts) que devuelve la matriz Q a partir

```python
import numpy as np
```

```python
# Esta función construye la matriz Q a partir de las relación entre puntos del tablero y puntos de la imagen.
# objPts (U, V): coordenadas reales de las esquinas del tablero,expresadas en milímetros.
# imgPts (u, v): coordenadas de esas mismas esquinas en la fotografía, expresadas en píxeles.

# Cada relación genera dos filas de Q, siguiendo la ecuación 11 del enunciado.

def obtenerQ2d(imgPts,objPts):
  matriz_Q=[] #Armo mi matriz Q en donde voy a guardar las filas creadas

  for i in range(len(objPts)) #Separo las coordenadas y aplico el algoritmo
    U= objPts[i][0]
    V= objPts[i][1]
    u= imgPts[i][0]
    v= imgPts[i][1]

    fila_1=[U, V, 1, 0, 0, 0, -u*U, -u*V, -u]
    fila_2=[0, 0, 0, U, V, 1, -v*V, -v*U, -v]

    matriz_Q.append(fila_1)
    matriz_Q.append(fila_2)

  return np.array(matriz_Q)
```
